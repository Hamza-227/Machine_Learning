# Scikit_learn_basics

[Scikit-learn Documentation](https://scikit-learn.org/stable/)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# simple linear regression
from sklearn.linear_model import LinearRegression
```

```python
# load a tips data 
df  = sns.load_dataset('tips')
df.head()
```

```python
# create a regression line plot using seaborn
sns.lmplot(x='total_bill', y='tip', data=df, height=6, aspect=1.5) 
```

```python
# Select the features column from df
X = df[['total_bill']]
Y = df[['tip']]
```

```python
# load and call linear regression model
model = LinearRegression()
# fit or training the model
model.fit(X, Y)
```

```python
# lets predict the modle from uunknown data like total_bill = 50
model.predict([[100]])
```

# Multiple Linear Regression

In which we have multiple feaatures and single label output so for tip data we use mutiple features like total_bill and size(number of people)

```python
# load the data
df = sns.load_dataset('tips')
# select the feature column from df
X = df[['total_bill', 'size']]
# select the target column from df
Y = df['tip']
# load/call the linear regression model
model = LinearRegression()
# fit the model means train the model
model.fit(X, Y)
# lets predict the tip for an unknown total_bill
model.predict([[35.527,2]])
```

now we use 3 columns as features like total_bill size and smoker
but smoker column is categorical so we need to convert it into numeric

```python
df['smoker_num']=df['smoker'].map({'Yes':1, 'No':0})
```

```python
df.head()
```

```python
# select the feature column from df
X = df[['total_bill', 'size','smoker_num']]
# select the target column from df
Y = df['tip']
# load/call the linear regression model
model = LinearRegression()
# fit the model means train the model
model.fit(X, Y)
# lets predict the tip for an unknown total_bill
model.predict([[35.527,2,1]])
```
